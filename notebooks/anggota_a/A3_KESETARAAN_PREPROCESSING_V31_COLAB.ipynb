## 3. Lihat gap dan sampel tidak identik
Mode `known_source` mereproduksi A2; mode `unknown_source` mendekati kondisi penggunaan umum; `old_streamlit` adalah fungsi lama. Perlu lihat sumber dan pola perbedaan sebelum revisi aplikasi.

```python
import csv
from collections import Counter
with open('HASIL_A3/AUDIT_KESETARAAN_A3_500_PER_SUMBER.csv',encoding='utf-8-sig',newline='') as f:
    mismatches=[r for r in csv.DictReader(f) if r['match_unknown']=='0' or r['match_old_streamlit']=='0']
print('Jumlah sampel berbeda dari salah satu mode inferensi:',len(mismatches))
for r in mismatches[:8]: print(r['raw_id'],r['source'],'unknown',r['match_unknown'],'old',r['match_old_streamlit'])
```
